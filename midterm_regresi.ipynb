```python
import pandas as pd
import os

# --- 1. SETUP LOAD DATA ---
filename = 'midterm-regresi-dataset.csv'
clean_filename = 'midterm_regresi_clean.csv'

if os.path.exists(filename):
    print("Loading data asli...")
    # Header=None sangat PENTING karena baris pertama adalah data
    df = pd.read_csv(filename, header=None)
    
    # Rename kolom agar lebih rapi (Year + 90 Fitur)
    # Kolom 0 adalah Target (Tahun), 1-90 adalah Fitur Audio
    df.columns = ["Year"] + [f"Feature_{i}" for i in range(1, df.shape[1])]
    
    print(f"Shape Awal: {df.shape}")
else:
    raise FileNotFoundError(f"File {filename} tidak ditemukan! Pastikan path sudah benar.")

# --- 2. CEK & BERSIHKAN DATA ---
def clean_regression_data(df):
    print("\n--- Memulai Pengecekan Data Kotor ---")
    initial_rows = df.shape[0]
    
    # A. Cek Missing Values (NaN)
    null_counts = df.isnull().sum().sum()
    if null_counts > 0:
        print(f"-> Ditemukan {null_counts} nilai kosong (NaN).")
        # Strategi: Drop baris (karena data banyak) atau Impute
        # Untuk regresi audio presisi, drop biasanya lebih aman jika jumlahnya sedikit
        df = df.dropna()
        print("   [FIX] Baris dengan NaN telah dihapus.")
    else:
        print("-> Tidak ada Missing Values (Aman).")
        
    # B. Cek Duplikat
    # Baris yang sama persis biasanya tidak berguna dan bikin bias
    duplicates = df.duplicated().sum()
    if duplicates > 0:
        print(f"-> Ditemukan {duplicates} baris duplikat.")
        df = df.drop_duplicates()
        print("   [FIX] Duplikat telah dihapus.")
    else:
        print("-> Tidak ada Duplikat (Aman).")
        
    # C. Cek Data Target Aneh (Opsional tapi penting untuk Regresi Tahun)
    # Misal: Apakah ada tahun 0 atau tahun 9999?
    invalid_years = df[(df["Year"] < 1900) | (df["Year"] > 2025)]
    if not invalid_years.empty:
        print(f"-> Ditemukan {len(invalid_years)} tahun yang tidak valid (di luar 1900-2025).")
        df = df[(df["Year"] >= 1900) & (df["Year"] <= 2025)]
        print("   [FIX] Tahun tidak valid dihapus.")
    
    # Ringkasan Akhir
    final_rows = df.shape[0]
    deleted = initial_rows - final_rows
    print(f"\n--- Selesai. Data dihapus: {deleted} baris ({deleted/initial_rows:.2%}) ---")
    print(f"Shape Akhir: {df.shape}")
    
    return df

# Eksekusi Pembersihan
df_clean = clean_regression_data(df)

# --- 3. SIMPAN DATA BERSIH ---
# Kita simpan agar tahap EDA & Modeling nanti pakai yang ini
print(f"\nMenyimpan data bersih ke '{clean_filename}'...")
df_clean.to_csv(clean_filename, index=False)
print("Sukses! Siap untuk EDA.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import os

# 1. LOAD DATA
filename = 'midterm_regresi_clean.csv'
if os.path.exists(filename):
    df = pd.read_csv(filename)
    print("Data Loaded.")
else:
    raise FileNotFoundError("File midterm_regresi_clean.csv tidak ditemukan!")

# 2. SPLIT FITUR (X) & TARGET (y)
X = df.drop(columns=['Year'])
y = df['Year']

print(f"Dimensi X: {X.shape}")
print(f"Dimensi y: {y.shape}")

# 3. SCALING (STANDARD SCALER)
# Penting agar Feature_47 (ribuan) tidak mendominasi Feature_1 (satuan)
print("\nMelakukan Scaling (StandardScaler)...")
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Ubah kembali ke DataFrame agar nama kolom tetap ada
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# 4. VALIDASI VISUAL: BOXPLOT SETELAH SCALING
# Kita lihat lagi 3 fitur yang tadi "rusak" grafiknya
top_3_features = ['Feature_1', 'Feature_7', 'Feature_47']

plt.figure(figsize=(10, 5))
# Ambil sampel 10.000 data saja biar plotting cepat
df_sample = X_scaled_df[top_3_features].sample(10000, random_state=42)
df_melt = df_sample.melt(var_name='Feature', value_name='Scaled Value')

sns.boxplot(data=df_melt, x='Feature', y='Scaled Value', palette="viridis")
plt.title("Boxplot Setelah Scaling (Rentang Nilai Sudah Setara)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 5. SPLIT TRAIN & TEST (80% Train, 20% Test)
print("\nMembagi Data Train & Test...")
X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, test_size=0.2, random_state=42)

print(f"Shape X_train: {X_train.shape}")
print(f"Shape X_test : {X_test.shape}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import xgboost as xgb
import numpy as np
import time

print("--- MEMULAI TRAINING MODEL REGRESI ---")

# ==========================================
# MODEL 1: LINEAR REGRESSION (Baseline)
# ==========================================
print("\n1. Training Linear Regression...")
lr = LinearRegression()
lr.fit(X_train, y_train)

# Prediksi & Evaluasi
y_pred_lr = lr.predict(X_test)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
r2_lr = r2_score(y_test, y_pred_lr)

print(f"-> RMSE (Linear): {rmse_lr:.4f} (Semakin kecil semakin bagus)")
print(f"-> R2 Score     : {r2_lr:.4f} (Mendekati 1.0 semakin bagus)")


# ==========================================
# MODEL 2: XGBOOST REGRESSOR (Advanced)
# ==========================================
print("\n2. Training XGBoost Regressor...")
start_time = time.time()

# Konfigurasi XGBoost untuk Regresi
xgbr = xgb.XGBRegressor(
    n_estimators=500,       # Jumlah pohon
    learning_rate=0.05,     # Kecepatan belajar
    max_depth=9,            # Kedalaman pohon
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,              # Pakai semua CPU
    objective='reg:squarederror' # Objective khusus regresi
)

# Training (tanpa eval_set dulu biar clean outputnya)
xgbr.fit(X_train, y_train)

# Prediksi & Evaluasi
y_pred_xgb = xgbr.predict(X_test)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb = r2_score(y_test, y_pred_xgb)

print(f"-> Selesai dalam {time.time() - start_time:.2f} detik.")
print(f"-> RMSE (XGBoost): {rmse_xgb:.4f}")
print(f"-> R2 Score      : {r2_xgb:.4f}")


# ==========================================
# KESIMPULAN
# ==========================================
print("\n" + "="*40)
print("     HASIL AKHIR (RMSE Terkecil Menang)")
print("="*40)
print(f"1. Linear Regression : {rmse_lr:.4f}")
print(f"2. XGBoost Regressor : {rmse_xgb:.4f}")
print("="*40)

if rmse_xgb < rmse_lr:
    print("Kesimpulan: XGBoost LEBIH BAIK daripada Linear Regression.")
else:
    print("Kesimpulan: Linear Regression LEBIH BAIK (Jarang terjadi).")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import xgboost as xgb
import joblib # Untuk menyimpan model

print("--- FINAL PROCESSING DENGAN MODEL TERBAIK (XGBoost) ---")

# 1. Pastikan Model XGBoost (xgbr) sudah dilatih
# Jika variabel hilang, kita train ulang sebentar
if 'xgbr' not in locals():
    print("Model tidak ditemukan di memori. Melatih ulang XGBoost...")
    xgbr = xgb.XGBRegressor(
        n_estimators=500, learning_rate=0.05, max_depth=9,
        subsample=0.8, colsample_bytree=0.8, random_state=42,
        n_jobs=-1, objective='reg:squarederror'
    )
    xgbr.fit(X_train, y_train)

# 2. Buat DataFrame Perbandingan (Actual vs Predicted)
# Kita gunakan data X_test yang belum pernah dilihat model saat training
y_final_pred = xgbr.predict(X_test)

results_df = pd.DataFrame({
    'Actual_Year': y_test,
    'Predicted_Year': y_final_pred
})

# Hitung Error (Selisih)
results_df['Error_Year'] = results_df['Actual_Year'] - results_df['Predicted_Year']
results_df['Abs_Error'] = results_df['Error_Year'].abs()

# Tampilkan 10 Prediksi Terbaik (Error paling kecil)
print("\n[TOP 10 PREDIKSI PALING AKURAT]")
print(results_df.sort_values(by='Abs_Error').head(10))

# Tampilkan 10 Prediksi Terburuk (Error paling besar)
print("\n[TOP 10 PREDIKSI PALING MELESET]")
print(results_df.sort_values(by='Abs_Error', ascending=False).head(10))


# 3. VISUALISASI PERFORMA MODEL
plt.figure(figsize=(14, 6))

# Plot A: Scatter Plot Actual vs Predicted
plt.subplot(1, 2, 1)
sns.scatterplot(x=results_df['Actual_Year'], y=results_df['Predicted_Year'], alpha=0.3, color='blue')
# Garis diagonal sempurna (jika prediksi 100% benar)
plt.plot([results_df['Actual_Year'].min(), results_df['Actual_Year'].max()],
         [results_df['Actual_Year'].min(), results_df['Actual_Year'].max()],
         'r--', lw=2, label='Perfect Prediction')
plt.title('Actual vs Predicted Year')
plt.xlabel('Tahun Asli')
plt.ylabel('Tahun Prediksi')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Plot B: Distribusi Error (Residuals)
plt.subplot(1, 2, 2)
sns.histplot(results_df['Error_Year'], bins=50, kde=True, color='orange')
plt.title('Distribusi Error (Seberapa Meleset Tebakannya?)')
plt.xlabel('Selisih Tahun (Actual - Prediksi)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# 4. SIMPAN MODEL TERBAIK
model_filename = 'best_model_xgboost_regresi.pkl'
joblib.dump(xgbr, model_filename)
print(f"\nModel terbaik berhasil disimpan sebagai '{model_filename}'.")
print("Anda bisa meload-nya kembali nanti dengan: model = joblib.load('nama_file')")
```
